```python
import numpy as np
import math
from sklearn.metrics import mean_squared_error
```

```python
# Set random seed for reproducibility
np.random.seed(26)
```

```python
# Import from DeepValve package
from deepvalve import Network, Layer, Neuron, ActivationFunction, LossFunction
```

## Testing DeepValve Library

Create and test a neural network using the deepvalve package.

```python
# Create a network with Huber loss
nw = Network(loss_name='huber')

# Add layers
nw.add_layer(num_neurons=2, num_inputs=2, activation='relu') 
nw.add_layer(num_neurons=3, activation='tanh')
nw.add_layer(num_neurons=1, activation='linear', is_output=True)

print("Network Architecture:")
for i, layer in enumerate(nw.layers):
    print(f"Layer {i+1}: {layer}")
    for j, neuron in enumerate(layer.neurons):
        print(f"  Neuron {j+1}: {neuron}")
```

```python
# Test forward pass with single input
input_data = np.array([0.5, -1.5])
output = nw.predict(input_data)
print(f"Network output: {output}")

# Test layer by layer
print("\nLayer by layer:")
layer1_output = nw.layers[0].forward(input_data)
print(f"Layer 1 output: {layer1_output}")

layer2_output = nw.layers[1].forward(layer1_output)
print(f"Layer 2 output: {layer2_output}")

layer3_output = nw.layers[2].forward(layer2_output)
print(f"Layer 3 (final) output: {layer3_output}")
```

```python
# Check weight and bias shapes
print("Weight and Bias Shapes:")
for i, layer in enumerate(nw.layers):
    for j, neuron in enumerate(layer.neurons):
        print(f"Layer {i+1}, Neuron {j+1} - Weights shape: {neuron.weights.shape}, Bias shape: {np.array(neuron.bias).shape}")
```

```python
# Test training
X_train = np.array([[0.1, 0.2], [0.3, 0.4]])
y_train = np.array([0.3, 0.7])

nw.fit(X_train, y_train, epochs=500, learning_rate=0.01)
```

```python
# Test prediction on training data
predictions = nw.predict(X_train)
print(f"Predictions: {predictions}")
print(f"Ground Truth: {y_train}")
print(f"MSE: {mean_squared_error(y_train, predictions)}")
```
